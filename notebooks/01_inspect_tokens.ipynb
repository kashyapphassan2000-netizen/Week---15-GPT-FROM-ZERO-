# 01 - inspect tokens (Day 2 gate)
Catches catastrophic tokenization bugs before a paid training run.

```python
import collections
import matplotlib.pyplot as plt
import numpy as np
import tiktoken

enc = tiktoken.get_encoding("r50k_base")
arr = np.memmap("../data/tokens/train.bin", dtype=np.uint16, mode="r")
print("train tokens:", len(arr))
```

```python
# top tokens
sample = arr[:1_000_000]
for tid, cnt in collections.Counter(sample.tolist()).most_common(20):
    print(f"{tid:>6}  {cnt:>7}  {enc.decode([tid])!r}")
```

```python
# doc length distribution (EOT=50256)
ends = np.where(arr[:5_000_000] == enc.eot_token)[0]
doc_lens = np.diff(ends)
print("docs:", len(doc_lens), "median:", int(np.median(doc_lens)),
      "p95:", int(np.percentile(doc_lens, 95)))
```

```python
# histogram
plt.hist(doc_lens, bins=80, range=(0, 5000))
plt.xlabel("tokens / doc"); plt.ylabel("count"); plt.show()
```
